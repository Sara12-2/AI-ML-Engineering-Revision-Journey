### Chapter 2 – Conditional Statements (if, elif, else)

An if statement is a decision-making statement that executes a block of code only when a specified condition evaluates to True.

#### Largest of 2 numbers 

```python
a = int(input("First Number: "))
b = int(input("Second Number: "))

if a > b:
    print(f"{a} is larger.")
else:
    print(f"{b} is larger.")
```
