### LeetCode #1929 — Concatenation of Array

Difficulty: Easy

#### Problem Statement

Given an integer array nums of length n, create an array ans of length 2n where:

ans[i] == nums[i] for 0 <= i < n
ans[i + n] == nums[i] for 0 <= i < n

In simple words:

Return an array that contains nums twice in the same order.

```python
class Solution:
    def getConcatenation(self, nums):
        ans = []

        for num in nums:
            ans.append(num)

        for num in nums:
            ans.append(num)

        return ans
    # Time complexity O(n), where n is the length of the input list nums. This is because we iterate through the list twice, resulting in a linear time complexity.
    #space complexity O(n), where n is the length of the input list nums. The space complexity is linear because we create a new list ans that contains twice the number of elements as the input list nums.
```

### LeetCode #1480 — Running Sum of 1D Array

```python
nums = [1,2,3,4]

ans = []

total = 0

for num in nums:
    total += num
    ans.append(total)

print(ans)
#Time complexity O(n), where n is the length of the input list nums. This is because we iterate through the list once, resulting in a linear time complexity.
#Space complexity O(n), where n is the length of the input list nums. The space complexity is linear because we create a new list ans that contains the same number of elements as the input list nums.
```

### LeetCode #1672 – Richest Customer Wealth

Difficulty: Easy ⭐

Topics: Arrays, Loops, Matrix

#### 📌 Problem Statement (English)

You are given an m × n integer matrix called accounts.

Each row represents a customer.
Each column represents the amount of money the customer has in a particular bank.

Your task is to return the wealth of the richest customer.

A customer's wealth is the sum of all money in their bank accounts.

Visit each customer (row).
Calculate their total wealth.
Compare it with the current maximum wealth.
Update if it's larger.
Return the maximum.

```python
accounts = [
    [1,5],
    [7,3],
    [3,5]
]

max_wealth = 0

for customer in accounts:

    total = 0

    for money in customer:
        total += money

    if total > max_wealth:
        max_wealth = total

print(max_wealth)
```

### Using Sum

```python
accounts = [
    [1,5],
    [7,3],
    [3,5]
]

max_wealth = 0

for customer in accounts:

    total = sum(customer)

    if total > max_wealth:
        max_wealth = total

print(max_wealth)
```
