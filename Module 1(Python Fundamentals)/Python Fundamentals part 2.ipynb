### Strings

A String is a sequence of characters.A string is an ordered sequence of characters. In Python, strings are immutable, meaning they cannot be modified after creation.

```python
name = "Sara"

city = 'Layyah'

paragraph = """I am learning
Python Programming."""

print(name)
print(city)
print(paragraph)
```

```python
text = "Artificial Intelligence"

count = 0

for ch in text.lower():

    if ch in "aeiou":
        count += 1

print("Total Vowels =", count)
```

```python
first = "Sara"

last = "Manzoor"

year = "2005"

username = first.lower() + "_" + last.lower() + year

print(username)

print("Length =", len(username))

print("Reverse =", username[::-1])
```

```python
skills = "Python, Flask, Docker, Git, MySQL"

skill_list = []

for skill in skills.split(","):
    skill_list.append(skill.strip().upper())

print("Skills List:")

for skill in skill_list:
    print(skill)

print("Total Skills =", len(skill_list))

print("Joined =", " | ".join(skill_list))
```
