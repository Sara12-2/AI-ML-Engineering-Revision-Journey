### 20. CSV Basics

CSV = Comma Separated Values

```python
import csv

with open("students.csv","w",newline="") as file:

    writer = csv.writer(file)

    writer.writerow(["Name","Age"])

    writer.writerow(["Sara",21])

    writer.writerow(["Ali",22])
```

### reading csv

```python
import csv

with open("students.csv","r") as file:

    reader = csv.reader(file)

    for row in reader:

        print(row)
```

## Exception Handling

### 1. What is an Exception?

An exception is a runtime error that interrupts the normal execution of a program.

### 2. Why Exception Handling?

```python
try:

    num = int(input("Enter Number: "))

    print(100 / num)

except:

    print("Something went wrong.")
```

Specific Exception

Better practice.

```python
try:

    num = int(input("Enter Number: "))

    print(100 / num)

except ZeroDivisionError:

    print("Cannot divide by zero.")
```

```python
try:

    age = int(input("Enter Age: "))

except ValueError:

    print("Please enter numbers only.")
```

### Multiple Exceptions

```python
try:

    num = int(input("Enter Number: "))

    print(100 / num)

except ValueError:

    print("Invalid Number")

except ZeroDivisionError:

    print("Cannot Divide by Zero")
```

One except for Multiple Errors

```python
try:

    num = int(input())

    print(100 / num)

except (ValueError, ZeroDivisionError):

    print("Invalid Input")
```

Exception as e

Professional way.

```python
try:

    num = int(input())

    print(100 / num)

except Exception as e:

    print(e)
```

#### Else Block

Runs only if no exception occurs.

```python
try:

    num = int(input())

except ValueError:

    print("Wrong Input")

else:

    print("Everything OK")
```

#### finally Block

Always executes.

```python
try:

    print(10 / 2)

except:

    print("Error")

finally:

    print("Program Ended")
```

#### Raising exceptions

```python
age = int(input())

if age < 18:

    raise ValueError("Age must be at least 18.")

print("Eligible")
```

### Custom Exception

```python
class InvalidAgeError(Exception):

    pass

age = int(input())

if age < 18:

    raise InvalidAgeError("Not Eligible")
```
